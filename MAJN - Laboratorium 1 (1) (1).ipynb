**Zadanie 7**. Załóżmy, że chcesz stworzyć chatbota do autonomicznej taksówki, który przyjmuje zlecenia przewozu. Utwórz wyrażenie regularne, które zwróci cel podróży na podstawie łańcucha znaków `zawieź mnie do [cel podróży]`. Następnie spróbuj rozszerzyć funkcjonalność przez uwzględnienie:
- synonimów słowa `zawieź` (np. `Jedź do ZOO`);
- możliwości wystąpienia słów takich jak `proszę`, `cześć`, `szybko`, itp.;
- określenia dodatkowo na końcu łańcucha punktu pośredniego na trasie, np. `Zawieź mnie do Gdańska przez Łódź`. W takim wypadku wyrażenie powinno też zwracać punkt pośredni oprócz celu podróży;
- możliwości kilku celów pośrednich oddzielonych przecinkami np. `Zawieź mnie do Lizbony przez Berlin, Paryż, Madryt` (tutaj także wyrażenie powinno zwracać każdy punkt na trasie);
- własnych pomysłów możliwych modyfikacji polecenia.
  
Przetestuj prawidłowe rozpoznawanie różnych zleceń.

```python
# Wersja podstawowa
import re

def zamowienie_podstawowe(text):
    m = re.search(r'zawieź mnie do\s+(.+)', text, re.IGNORECASE)
    return m.group(1) if m else None

print(zamowienie_podstawowe("Zawieź mnie do Gdańska"))
# Gdańska
```

```python
#Wersja rozszerzona
import re

def zamowienie(text):
    # synonimy: zawieź, jedź, pojedź, zawieźć, dostarcz, wieź
    # opcjonalne wtrącenia: proszę, cześć, szybko, no, hej, bardzo proszę
    wzorzec = (
        r'^\s*'
        r'(?:proszę\s+|cześć\s+|hej\s+|szybko\s+|no\s+|bardzo proszę\s+)*'
        r'(?:zawieź|zawieźć|jedź|pojedź|dostarcz|wieź)\s+'
        r'(?:mnie\s+)?'
        r'do\s+'
        r'(?P<cel>[A-ZĄĆĘŁŃÓŚŹŻ][\wąćęłńóśźż-]*(?:\s+[A-ZĄĆĘŁŃÓŚŹŻ][\wąćęłńóśźż-]*)*)'
        r'(?:\s+przez\s+(?P<posrednie>.+?))?'
        r'\s*[.!]?\s*$'
    )
    m = re.match(wzorzec, text, re.IGNORECASE)
    if not m:
        return None
    
    cel = m.group('cel').strip()
    posrednie_raw = m.group('posrednie')
    
    if posrednie_raw:
        posrednie = [p.strip() for p in posrednie_raw.split(',') if p.strip()]
    else:
        posrednie = []
    
    return {'cel': cel, 'posrednie': posrednie}

# Testy
testy = [
    "Zawieź mnie do Gdańska",
    "Jedź do ZOO",
    "Proszę zawieź mnie szybko do Krakowa",
    "Cześć, zawieź mnie do Gdańska przez Łódź",
    "Zawieź mnie do Lizbony przez Berlin, Paryż, Madryt",
    "Pojedź mnie do Warszawy przez Radom, Kielce",
    "Dostarcz mnie do Poznania",
    "Wieź mnie do Wrocławia przez Opole, Katowice, Kraków",
]

for t in testy:
    print(f"{t!r:60} -> {zamowienie(t)}")
```

**Zadanie 8**. Tam, gdzie zadanie 7 okazało się zbyt trudne, wykorzystaj wybrane narzędzie sztucznej inteligencji do określenia właściwego wyrażenia regularnego. Następnie: 
1. spróbuj znaleźć poprawne zlecenie, które nie jest rozpoznawane,
2. jeżeli uda się znaleźć takie zlecenie, ponownie wykorzystaj sztuczną inteligencję do poprawienia odpowiedzi lub popraw wyrażenie samodzielnie.
  
Powtórz punkty 1. i 2. jeszcze dwa razy.

```python
import re

def zamowienie_final(text):
    regex = (
        r'^\s*'
        r'(?:(?:proszę|cześć|hej|szybko|no|bardzo proszę)\s*[,!]?\s*)*'
        r'(?:zawieź|zawieźć|jedź|pojedź|dostarcz|wieź)\s+'
        r'(?:mnie\s+)?'
        r'(?:do|na)\s+'
        r'(?P<cel>[A-ZĄĆĘŁŃÓŚŹŻ][\wąćęłńóśźż-]*(?:\s+[A-ZĄĆĘŁŃÓŚŹŻ][\wąćęłńóśźż-]*)*)'
        r'(?:\s*,?\s*przez\s+(?P<posrednie>.+?))?'
        r'\s*[.!]?\s*$'
    )
    m = re.match(regex, text, re.IGNORECASE)
    if not m:
        return None
    cel = m.group('cel').strip()
    posrednie_raw = m.group('posrednie')
    posrednie = re.split(r'\s*,\s*|\s+i\s+', posrednie_raw) if posrednie_raw else []
    posrednie = [p.strip() for p in posrednie if p.strip()]
    return {'cel': cel, 'posrednie': posrednie}

# Testy finalne
testy = [
    "Zawieź mnie do Gdańska",
    "Jedź na lotnisko",
    "Cześć, zawieź mnie do Gdańska przez Łódź.",
    "Zawieź mnie do Lizbony przez Berlin, Paryż, Madryt",
    "Zawieź mnie do Warszawy, przez Radom i Kielce",
    "Proszę, szybko jedź mnie do Krakowa przez Katowice, Opole i Wrocław!",
]
for t in testy:
    print(f"{t!r:65} -> {zamowienie_final(t)}")
```
